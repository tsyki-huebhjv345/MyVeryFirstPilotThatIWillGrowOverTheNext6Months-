```python
from qiskit.quantum_info import Statevector
from numpy import sqrt
from qiskit.visualization import array_to_latex
from qiskit.visualization import plot_histogram

u = Statevector ([1 / sqrt(2), 1 / sqrt(2)])
v = Statevector([(1 + 2.0j) / 3, -2 / 3])
w = Statevector([1 / 3, 2 / 3])

display(array_to_latex([u,v,w]))

display(u.draw("text"))
display(u.draw("latex"))
print(u.draw("latex_source"))




statistics = v.sample_counts(1000)
plot_histogram(statistics) #law of large numbers - the more you run the program the more the numbers will get
#closer to their exact probability. run the code multiple times to see how the numbers change
```
