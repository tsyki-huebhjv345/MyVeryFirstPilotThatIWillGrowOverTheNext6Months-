```python
from qiskit.quantum_info import Statevector
from numpy import sqrt
from qiskit.visualization import array_to_latex

u = Statevector ([1 / sqrt(2), 1 / sqrt(2)])
v = Statevector([(1 + 2.0j) / 3, -2 / 3])
w = Statevector([1 / 3, 2 / 3])

display(array_to_latex([u,v,w]))

display(u.draw("text"))
display(u.draw("latex"))
print(u.draw("latex_source"))

display(u.is_valid())
display(w.is_valid())

outcome, state = v.measure()
print(f"Measured: {outcome}\nPost-measurement state:")
display(state.draw("latex")) # if your measurements are 0  then: in order to normalise this we just square up, sum up and take the root of the nominator getting the length this way, and then dividing the initial, unnormalised vector by that lenght. the denominators will cancel out leaving us with the fraction that mathematically guarantees to be 1! run this code multiple times and the measurements will change!
#Statevector will throw an error if the measure method is applied to an invalid quantum state vector.
```
